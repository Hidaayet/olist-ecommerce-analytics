# 01 - Data inspection
Profile of the 9 raw tables. Findings are summarised in `docs/decisions_log.md`.

```python
import pandas as pd
from pathlib import Path

RAW = Path("../data/raw")
tables = {f.stem: pd.read_csv(f) for f in sorted(RAW.glob("*.csv"))}

for name, df in tables.items():
    print(f"\n{'='*60}\n{name}  |  {df.shape[0]:,} rows x {df.shape[1]} cols")
    summary = pd.DataFrame({
        "dtype": df.dtypes,
        "nulls": df.isna().sum(),
        "null_%": (df.isna().mean() * 100).round(2),
        "unique": df.nunique(),
    })
    print(summary)
    print(f"Fully duplicated rows: {df.duplicated().sum():,}")
```

```python
o, items, pay, rev, cust = (tables[k] for k in [
    "olist_orders_dataset", "olist_order_items_dataset",
    "olist_order_payments_dataset", "olist_order_reviews_dataset",
    "olist_customers_dataset"])

print("orders PK unique:", o.order_id.is_unique)
print("customer_id unique in orders:", o.customer_id.is_unique)
print("customer_unique_id vs customer_id:", cust.customer_unique_id.nunique(), "vs", cust.customer_id.nunique())
print("orders with >1 review:", (rev.order_id.value_counts() > 1).sum())
print("orders with >1 payment row:", (pay.order_id.value_counts() > 1).sum())
print(o.order_status.value_counts())
```
